```python
# Logic: if condition(arr[0][0]%2==0) then np.square else arr[0][0]
```

```python
# https://stackoverflow.com/questions/42540224/conditional-operations-on-numpy-arrays
```

```python
# Since numpy is used to perform on entire array s we cannot use if else condition 
```

```python
# using np.where() ---> np.where(condition; if true what to do, if false what to do)
```

```python
np.where(array_greeshma%2==0, np.square(array_greeshma), array_greeshma)
```

```python
resulting_array=np.where(array_greeshma%2==0, np.square(array_greeshma), array_greeshma)
```

```python
print(resulting_array)
```
